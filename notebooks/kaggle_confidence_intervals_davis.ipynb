# ColdSite-DTI — how precisely is each number measured? (Kaggle, T4 x2)

Every table in the audit carries a spread over three seeds. That says how much retraining
moves a number; it does not say how precisely a cell measures it. At the cold levels a
cell rests on 68 proteins, and against the drug-specific ground truth on as few as 6
pairs — where a precision@10 of 0.133 is one lucky protein and 0.133 on 349 proteins is a
result. Without an interval those look identical in the table.

This notebook resamples **proteins** (`src/evaluation/bootstrap_ci.py`): draw n proteins
with replacement from the n a cell scored, recompute the mean, 10,000 times, and report
the 2.5th and 97.5th percentiles. A protein that was scored under several seeds enters
the resample **with all of its seeds at once**, so the interval carries protein sampling
and seed noise together, and one protein measured three times is not mistaken for three
proteins.

**The intervals themselves cost half a second.** What costs time is re-scoring the
ladders so they record a value per protein, which the earlier runs did not keep -- a mean
cannot be given an interval after the fact. That is all this notebook spends GPU on.

| arm | jobs | roughly |
|---|---|---|
| attention, UniProt + KLIFS | 18 | ~15 min |
| drug-specific (own, swapped, paired) | 18 | ~10 min |
| integrated gradients, UniProt + KLIFS | 18 | **~1.5 h** (32 passes per pair) |

`RUN_IG` is off by default for that reason: turn it on when you want intervals on the IG
numbers too, and expect the run to take about two hours instead of twenty-five minutes.

## 5. Collect the trained cells — and verify every single one

Every attached dataset is searched for the 48 `(model, level, seed)` cells. For each cell
the **test AUROC in its results file must match the value this project verified on the
Mac**, to four decimals. That one check catches, without any judgement on your part:

- a checkpoint that was cut off mid-training (its results file is absent, or its number
  differs);
- MolTrans's invalid seeds 2 and 3 from the first grid, which trained as seed 1 (the
  vendored `models.py` reseeds torch on import) and whose AUROCs are seed 1's;
- an older copy of a cell that was later retrained;
- a dataset that is simply the wrong one.

If several attached datasets hold the same cell, the copy whose AUROC matches is the one
used, and its checkpoint is taken from the same folder. Nothing is guessed: if no copy
matches, the notebook stops and names the cell.

## 6. The runner

Each stage is a list of real command-line calls — the same commands the project runs on a
laptop — spread over the GPUs. Output is streamed with a `[GPU0 ...]` prefix so two
parallel jobs stay readable, a STATUS line prints every 10 minutes, and everything stops at
the deadline with its finished outputs intact.

```python
import re
import subprocess
import threading

STATUS_EVERY = 600


def py(module, *args):
    return ['python', '-u', '-m', module, *map(str, args)]


def run_parallel(queues, label):
    """queues: {gpu: [(name, cmd, expected_output), ...]}. True if the deadline cut in."""
    queues = {g: q for g, q in queues.items() if q}
    where = {g: {'name': '-', 'done': 0, 'total': len(q)} for g, q in queues.items()}
    state = {'deadline': False, 'failed': []}
    running = {}

    def worker(gpu, jobs):
        env = {**os.environ, 'CUDA_VISIBLE_DEVICES': str(gpu), 'PYTHONUNBUFFERED': '1'}
        with open(f'{WORK}/{label}_gpu{gpu}.log', 'a') as log:
            for name, cmd, expected in jobs:
                if time.time() > DEADLINE:
                    return
                if SKIP_EXISTING and expected and os.path.exists(expected):
                    print(f'[skip] GPU{gpu} {name} -- {os.path.basename(expected)} exists',
                          flush=True)
                    where[gpu]['done'] += 1
                    continue
                where[gpu]['name'] = name
                began = time.time()
                print(f'\n{"=" * 70}\n[GPU{gpu}] {name}\n{"=" * 70}', flush=True)
                proc = subprocess.Popen(cmd, env=env, stdout=subprocess.PIPE,
                                        stderr=subprocess.STDOUT, text=True, bufsize=1)
                running[gpu] = proc
                for line in proc.stdout:
                    if not re.match(r'\s*(test |train )?batch \d+/', line):   # drop batch spam
                        print(f'[GPU{gpu}] {line}', end='', flush=True)
                    log.write(line)
                code_ = proc.wait()
                where[gpu]['done'] += 1
                took = time.time() - began
                if code_ != 0 or (expected and not os.path.exists(expected)):
                    why = f'exited {code_}' if code_ else f'wrote no {os.path.basename(expected)}'
                    print(f'!! [GPU{gpu}] {name} FAILED ({why}, {took:.0f}s)', flush=True)
                    if not state['deadline']:
                        state['failed'].append(name)
                else:
                    print(f'   [GPU{gpu}] {name} done in {took:.0f}s', flush=True)

    threads = [threading.Thread(target=worker, args=(g, q), daemon=True)
               for g, q in queues.items()]
    for t in threads:
        t.start()
    last = 0.0
    while any(t.is_alive() for t in threads):
        if time.time() - last >= STATUS_EVERY:
            last = time.time()
            parts = [f'GPU{g}: {w["name"]} ({w["done"]}/{w["total"]})' for g, w in where.items()]
            print(f'\n=== STATUS {time.strftime("%H:%M")} | ' + ' | '.join(parts)
                  + f' | {hours_left():.1f} h left ===\n', flush=True)
        if time.time() > DEADLINE and not state['deadline']:
            state['deadline'] = True
            print('\n*** deadline: stopping so this commit saves its output. Re-run the '
                  'notebook with the output attached to finish the rest. ***\n', flush=True)
            for proc in list(running.values()):
                if proc.poll() is None:
                    proc.terminate()
        time.sleep(10)
    if state['failed']:
        print('\nFAILED: ' + ', '.join(state['failed']))
    return state['deadline']


def spread(jobs):
    """Deal jobs to the GPUs, heaviest first, so both finish at about the same time."""
    return {g: jobs[g::N_GPU] for g in range(N_GPU)}


COMMON = ['--split-root', 'data/splits', '--checkpoint-dir', RESULTS]
GT = f'data/{DATASET}_ground_truth_sites.json'
GT_KLIFS = f'data/{DATASET}_klifs_pocket_sites.json'
print('runner ready')
```

## 7. Re-score the ladders so each protein's value is kept

`run_ladder` records `per_protein` and the protein ids alongside every mean. Ladders
written before that change have only the mean, and `bootstrap_ci` reports those cells as
"predates per-protein scores" rather than quietly leaving them out.

```python
GT = f'data/{DATASET}_ground_truth_sites.json'
GT_KLIFS = f'data/{DATASET}_klifs_pocket_sites.json'
ARMS = {}

if RUN_ATTENTION:
    ARMS['uniprot'] = (GT, 1, f'{WORK}/ci_{DATASET}_uniprot', MODELS)
    ARMS['klifs'] = (GT_KLIFS, 1, f'{WORK}/ci_{DATASET}_klifs', MODELS)
if RUN_DRUG:
    for label, path in (('drug', f'data/{DATASET}_drug_sites.json'),
                        ('drug swapped', f'data/{DATASET}_drug_sites_swapped.json'),
                        ('drug paired', f'data/{DATASET}_drug_sites_paired.json')):
        assert os.path.exists(path), f'{path} is missing from this checkout'
        ARMS[label] = (path, 0, f'{WORK}/ci_{label.replace(" ", "_")}', MODELS)
if RUN_IG:
    ARMS['uniprot (IG)'] = (GT, 1, f'{WORK}/ci_ig_uniprot', [f'{m}_ig' for m in MODELS])
    ARMS['klifs (IG)'] = (GT_KLIFS, 1, f'{WORK}/ci_ig_klifs', [f'{m}_ig' for m in MODELS])

jobs = []
for label, (gt, cap, folder, models) in ARMS.items():
    os.makedirs(folder, exist_ok=True)
    for m in models:
        tag = f'{m}_{DATASET}' if m != 'coldsite_dti' else DATASET
        for s in SEEDS:
            jobs.append((f'{m} seed {s}: {label}',
                         py('src.evaluation.run_ladder', '--model', m, '--task', 'binary',
                            '--dataset', DATASET, '--seed', s, *COMMON,
                            '--ground-truth', gt, '--pairs-per-target', cap,
                            '--out-dir', folder, '--device', 'cuda'),
                         os.path.join(folder, f'ladder_{tag}_seed{s}.json')))

print(f'{len(jobs)} ladders to re-score across {len(ARMS)} arm(s)')
if jobs:
    cut = run_parallel(spread(jobs), 'ci')
    print()
    print('CUT SHORT by the deadline' if cut else 'every ladder re-scored')
```

## 8. The intervals

One table per arm. `n` is the number of proteins (or pairs) the cell rests on, which is
what sets the width: a cell of six cannot separate anything from chance, and the interval
is how the paper says so instead of quoting a bare point estimate.

```python
from src.evaluation.bootstrap_ci import cells as ci_cells, report

folders = {label: folder for label, (_gt, _cap, folder, _m) in ARMS.items()
           if os.path.isdir(folder)}
models_for = {label: models for label, (_gt, _cap, _f, models) in ARMS.items()}

rows = []
for label, folder in folders.items():
    rows += ci_cells({label: folder}, models_for[label], DATASET, SEEDS, K,
                     N_RESAMPLES, CONFIDENCE)

text = report(rows, K, CONFIDENCE)
print(text)
OUT_CI = f'{WORK}/ci_{DATASET}.md'
with open(OUT_CI, 'w') as handle:
    handle.write(text)
with open(OUT_CI.replace('.md', '.json'), 'w') as handle:
    json.dump(rows, handle, indent=1)
print()
print(f'Saved -> {OUT_CI}')
```
