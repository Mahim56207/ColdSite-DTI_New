# ColdSite-DTI — is it attention's fault, or the models'? (Kaggle, T4 x2)

Every number in the audit so far scores a model's **attention**. When attention misses the
binding site, two opposite things could be true: the attention is a poor report of a model
that does represent the site, or the model never learned it. No attention measurement can
tell them apart.

**Integrated gradients** can (`src/evaluation/integrated_gradients.py`). It attributes the
prediction to residues using only the trained weights and the gradient -- no
interpretability head -- along a path from the padding embedding (the same "no residue
here" the faithfulness masking uses) to the real one, explaining each model's own
`predict`. The variants `coldsite_dti_ig`, `hyperattentiondti_ig` and `moltrans_ig` read
the *same checkpoints* as the attention runs, so the comparison is two explanations of one
model rather than two models.

| if | then |
|---|---|
| IG finds the sites where attention does not | the information is in the model; **attention fails to report it** |
| neither finds them | the models do not represent the site — attention is not the problem |
| both find them | the claim holds, and the audit says so |

A first look on 20 proteins (DAVIS random, seed 1) favours the first row against the KLIFS
pocket -- ColdSite-DTI 0.160 -> 0.285 and HyperAttentionDTI 0.235 -> 0.490, chance ~0.14 --
while both stay at chance against UniProt's annotated residues. This notebook settles it on
the whole grid, 3 models x 3 seeds x 4 levels, against both ground truths.

- a dataset that is simply the wrong one.

If several attached datasets hold the same cell, the copy whose AUROC matches is the one

matches, the notebook stops and names the cell.

## 6. The runner

Each stage is a list of real command-line calls — the same commands the project runs on a
laptop — spread over the GPUs. Output is streamed with a `[GPU0 ...]` prefix so two
parallel jobs stay readable, a STATUS line prints every 10 minutes, and everything stops at
the deadline with its finished outputs intact.

```python
import re
import subprocess
import threading

STATUS_EVERY = 600


def py(module, *args):
    return ['python', '-u', '-m', module, *map(str, args)]


def run_parallel(queues, label):
    """queues: {gpu: [(name, cmd, expected_output), ...]}. True if the deadline cut in."""
    queues = {g: q for g, q in queues.items() if q}
    where = {g: {'name': '-', 'done': 0, 'total': len(q)} for g, q in queues.items()}
    state = {'deadline': False, 'failed': []}
    running = {}

    def worker(gpu, jobs):
        env = {**os.environ, 'CUDA_VISIBLE_DEVICES': str(gpu), 'PYTHONUNBUFFERED': '1'}
        with open(f'{WORK}/{label}_gpu{gpu}.log', 'a') as log:
            for name, cmd, expected in jobs:
                if time.time() > DEADLINE:
                    return
                if SKIP_EXISTING and expected and os.path.exists(expected):
                    print(f'[skip] GPU{gpu} {name} -- {os.path.basename(expected)} exists',
                          flush=True)
                    where[gpu]['done'] += 1
                    continue
                where[gpu]['name'] = name
                began = time.time()
                print(f'\n{"=" * 70}\n[GPU{gpu}] {name}\n{"=" * 70}', flush=True)
                proc = subprocess.Popen(cmd, env=env, stdout=subprocess.PIPE,
                                        stderr=subprocess.STDOUT, text=True, bufsize=1)
                running[gpu] = proc
                for line in proc.stdout:
                    if not re.match(r'\s*(test |train )?batch \d+/', line):   # drop batch spam
                        print(f'[GPU{gpu}] {line}', end='', flush=True)
                    log.write(line)
                code_ = proc.wait()
                where[gpu]['done'] += 1
                took = time.time() - began
                if code_ != 0 or (expected and not os.path.exists(expected)):
                    why = f'exited {code_}' if code_ else f'wrote no {os.path.basename(expected)}'
                    print(f'!! [GPU{gpu}] {name} FAILED ({why}, {took:.0f}s)', flush=True)
                    if not state['deadline']:
                        state['failed'].append(name)
                else:
                    print(f'   [GPU{gpu}] {name} done in {took:.0f}s', flush=True)

    threads = [threading.Thread(target=worker, args=(g, q), daemon=True)
               for g, q in queues.items()]
    for t in threads:
        t.start()
    last = 0.0
    while any(t.is_alive() for t in threads):
        if time.time() - last >= STATUS_EVERY:
            last = time.time()
            parts = [f'GPU{g}: {w["name"]} ({w["done"]}/{w["total"]})' for g, w in where.items()]
            print(f'\n=== STATUS {time.strftime("%H:%M")} | ' + ' | '.join(parts)
                  + f' | {hours_left():.1f} h left ===\n', flush=True)
        if time.time() > DEADLINE and not state['deadline']:
            state['deadline'] = True
            print('\n*** deadline: stopping so this commit saves its output. Re-run the '
                  'notebook with the output attached to finish the rest. ***\n', flush=True)
            for proc in list(running.values()):
                if proc.poll() is None:
                    proc.terminate()
        time.sleep(10)
    if state['failed']:
        print('\nFAILED: ' + ', '.join(state['failed']))
    return state['deadline']


def spread(jobs):
    """Deal jobs to the GPUs, heaviest first, so both finish at about the same time."""
    return {g: jobs[g::N_GPU] for g in range(N_GPU)}


COMMON = ['--split-root', 'data/splits', '--checkpoint-dir', RESULTS]
GT = f'data/{DATASET}_ground_truth_sites.json'
GT_KLIFS = f'data/{DATASET}_klifs_pocket_sites.json'
print('runner ready')
```

apart automatically (`ladder_moltrans_ig_davis_seed1.json`), so nothing overwrites the
attention results. MolTrans is the slow one and gets a GPU of its own where possible.

```python
OUT_IG = f'{WORK}/ig_{DATASET}'
OUT_IG_KLIFS = f'{OUT_IG}_klifs'
for d in (OUT_IG, OUT_IG_KLIFS):
    os.makedirs(d, exist_ok=True)

jobs = []
order = [m for m in ('moltrans', 'hyperattentiondti', 'coldsite_dti') if m in MODELS]
for m in order:
    variant = f'{m}_ig'
    for s in SEEDS:
        # ColdSite-DTI's attention is read from its own forward pass and its masking does
        # not go through residue_space, so run_faithfulness refuses its IG variant. The
        # ladder -- what part 6 is actually asking -- runs for all three.
        if RUN_FAITHFULNESS and m != 'coldsite_dti':
            jobs.append((f'{variant} seed {s}: faithfulness',
                         py('src.evaluation.run_faithfulness', '--model', variant,
                            '--dataset', DATASET, '--seed', s, '--task', 'binary',
                            *COMMON, '--out-dir', OUT_IG,
                            '--max-pairs', MAX_PAIRS, '--device', 'cuda'),
                         os.path.join(OUT_IG, f'faithfulness_{variant}_{DATASET}_seed{s}.json')))
        if RUN_UNIPROT:
            jobs.append((f'{variant} seed {s}: UniProt ladder',
                         py('src.evaluation.run_ladder', '--model', variant, '--task', 'binary',
                            '--dataset', DATASET, '--seed', s, *COMMON,
                            '--ground-truth', GT, '--out-dir', OUT_IG, '--device', 'cuda'),
                         os.path.join(OUT_IG, f'ladder_{variant}_{DATASET}_seed{s}.json')))
        if RUN_KLIFS:
            jobs.append((f'{variant} seed {s}: KLIFS ladder',
                         py('src.evaluation.run_ladder', '--model', variant, '--task', 'binary',
                            '--dataset', DATASET, '--seed', s, *COMMON,
                            '--ground-truth', GT_KLIFS, '--out-dir', OUT_IG_KLIFS,
                            '--device', 'cuda'),
                         os.path.join(OUT_IG_KLIFS, f'ladder_{variant}_{DATASET}_seed{s}.json')))

print(f'{len(jobs)} jobs')
if RUN_FAITHFULNESS and 'coldsite_dti' in MODELS:
    print('(no faithfulness job for coldsite_dti_ig: run_faithfulness refuses it, see '
          'the note in this cell -- its ladder still runs)')
if jobs:
    cut = run_parallel(spread(jobs), 'ig')
    print()
    print("CUT SHORT by the deadline -- re-run with this output attached"
          if cut else "every job finished")
```

## 9. Attention versus integrated gradients

The comparison this notebook exists for. Attention numbers come from the repository's own
analysis folders (computed in the attention runs); IG numbers from this run. Both are
precision@10 on the same cells, against the same ground truth, from the same weights.

```python
import statistics as st

import pandas as pd

K = '10'
LEVELS = ('random', 'cold_drug', 'cold_target', 'cold_pair')
ATTENTION = {'uniprot': f'results/analysis_{DATASET}_policyA',
             'klifs': f'results/analysis_{DATASET}_policyA_klifs'}
IG = {'uniprot': OUT_IG, 'klifs': OUT_IG_KLIFS}


def tag_of(model):
    return DATASET if model == 'coldsite_dti' else f'{model}_{DATASET}'


def at_k(folder, model, seed, level, field='precision_at_k'):
    path = os.path.join(folder, f'ladder_{tag_of(model)}_seed{seed}.json')
    if not os.path.exists(path):
        return None
    payload = json.load(open(path))
    return payload.get(level, {}).get('by_k', {}).get(K, {}).get(field)


def mean_of(folder, model, seed_list, level, field='precision_at_k'):
    # Plain arithmetic on purpose: statistics.* raises AttributeError on some value
    # types under the Python 3.12 Kaggle runs, and this cell runs after hours of work.
    values = [at_k(folder, model, s, level, field) for s in seed_list]
    values = [float(v) for v in values if v is not None]
    return sum(values) / len(values) if values else None


rows = []
for m in MODELS:
    for which in ('uniprot', 'klifs'):
        for level in LEVELS:
            attention = mean_of(ATTENTION[which], m, SEEDS, level)
            ig = mean_of(IG[which], f'{m}_ig', SEEDS, level)
            chance = mean_of(IG[which], f'{m}_ig', SEEDS, level, 'chance')
            rows.append({'model': m, 'ground truth': which, 'level': level,
                         'chance': f'{chance:.3f}' if chance else '—',
                         'attention': f'{attention:.3f}' if attention else '—',
                         'IG': f'{ig:.3f}' if ig else '—',
                         'IG - attention': f'{ig - attention:+.3f}'
                                           if (ig and attention) else '—'})

frame = pd.DataFrame(rows)
print(frame.to_string(index=False))
frame.to_csv(f'{WORK}/ig_vs_attention.csv', index=False)
print()
print(f"Saved -> {WORK}/ig_vs_attention.csv")
print("A dash under attention means that cell is not in the repository yet: "
      "the attention runs and this one are separate commits.")
```
