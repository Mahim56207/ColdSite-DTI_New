# ColdSite-DTI — is the verdict the model's, or our readout's? (Kaggle, T4 x2)

Between the tensor inside a network and the one number per residue that precision@k
scores, somebody chooses: which axis to average over, which layer to read, how to spread
a convolution position or a sub-word token across the residues it covers. Every number in
this audit rests on choices made once, early, and never varied. A reviewer is entitled to
ask whether the verdict describes the model or the choice.

This notebook scores the same checkpoints through readouts another author could reasonably
have picked (`src/evaluation/readout_variants.py`), each changing exactly one documented
choice:

| variant | what changes |
|---|---|
| `hyperattentiondti_maxchannel` | channel max instead of mean |
| `hyperattentiondti_receptive` | a convolution's weight spread over its receptive field, not placed on the centre residue |
| `moltrans_maxhead` | attention-head max instead of mean |
| `moltrans_firstlayer` | the first protein-encoder layer instead of the last |
| `coldsite_dti_selfattn` | the protein tower's own self-attention — which the forward pass computes and discards, and which cannot depend on the drug |

A first look on 25 proteins says the readouts pick **almost disjoint residues**: 2-12%
overlap in the top ten with the published readout (MolTrans's head max is the exception at
90%), and precision@10 against UniProt spanning 0.020-0.056. This notebook settles it on
the whole grid, against both ground truths, with all three seeds.

**What would count as a problem for the paper:** a variant that reverses a verdict — at
chance against UniProt's residues, above chance against the KLIFS pocket. A variant that
merely moves a number is the sensitivity the Limitations section promises to report.

## 6. The runner

Each stage is a list of real command-line calls — the same commands the project runs on a
laptop — spread over the GPUs. Output is streamed with a `[GPU0 ...]` prefix so two
parallel jobs stay readable, a STATUS line prints every 10 minutes, and everything stops at
the deadline with its finished outputs intact.

```python
import re
import subprocess
import threading

STATUS_EVERY = 600


def py(module, *args):
    return ['python', '-u', '-m', module, *map(str, args)]


def run_parallel(queues, label):
    """queues: {gpu: [(name, cmd, expected_output), ...]}. True if the deadline cut in."""
    queues = {g: q for g, q in queues.items() if q}
    where = {g: {'name': '-', 'done': 0, 'total': len(q)} for g, q in queues.items()}
    state = {'deadline': False, 'failed': []}
    running = {}

    def worker(gpu, jobs):
        env = {**os.environ, 'CUDA_VISIBLE_DEVICES': str(gpu), 'PYTHONUNBUFFERED': '1'}
        with open(f'{WORK}/{label}_gpu{gpu}.log', 'a') as log:
            for name, cmd, expected in jobs:
                if time.time() > DEADLINE:
                    return
                if SKIP_EXISTING and expected and os.path.exists(expected):
                    print(f'[skip] GPU{gpu} {name} -- {os.path.basename(expected)} exists',
                          flush=True)
                    where[gpu]['done'] += 1
                    continue
                where[gpu]['name'] = name
                began = time.time()
                print(f'\n{"=" * 70}\n[GPU{gpu}] {name}\n{"=" * 70}', flush=True)
                proc = subprocess.Popen(cmd, env=env, stdout=subprocess.PIPE,
                                        stderr=subprocess.STDOUT, text=True, bufsize=1)
                running[gpu] = proc
                for line in proc.stdout:
                    if not re.match(r'\s*(test |train )?batch \d+/', line):   # drop batch spam
                        print(f'[GPU{gpu}] {line}', end='', flush=True)
                    log.write(line)
                code_ = proc.wait()
                where[gpu]['done'] += 1
                took = time.time() - began
                if code_ != 0 or (expected and not os.path.exists(expected)):
                    why = f'exited {code_}' if code_ else f'wrote no {os.path.basename(expected)}'
                    print(f'!! [GPU{gpu}] {name} FAILED ({why}, {took:.0f}s)', flush=True)
                    if not state['deadline']:
                        state['failed'].append(name)
                else:
                    print(f'   [GPU{gpu}] {name} done in {took:.0f}s', flush=True)

    threads = [threading.Thread(target=worker, args=(g, q), daemon=True)
               for g, q in queues.items()]
    for t in threads:
        t.start()
    last = 0.0
    while any(t.is_alive() for t in threads):
        if time.time() - last >= STATUS_EVERY:
            last = time.time()
            parts = [f'GPU{g}: {w["name"]} ({w["done"]}/{w["total"]})' for g, w in where.items()]
            print(f'\n=== STATUS {time.strftime("%H:%M")} | ' + ' | '.join(parts)
                  + f' | {hours_left():.1f} h left ===\n', flush=True)
        if time.time() > DEADLINE and not state['deadline']:
            state['deadline'] = True
            print('\n*** deadline: stopping so this commit saves its output. Re-run the '
                  'notebook with the output attached to finish the rest. ***\n', flush=True)
            for proc in list(running.values()):
                if proc.poll() is None:
                    proc.terminate()
        time.sleep(10)
    if state['failed']:
        print('\nFAILED: ' + ', '.join(state['failed']))
    return state['deadline']


def spread(jobs):
    """Deal jobs to the GPUs, heaviest first, so both finish at about the same time."""
    return {g: jobs[g::N_GPU] for g in range(N_GPU)}


COMMON = ['--split-root', 'data/splits', '--checkpoint-dir', RESULTS]
GT = f'data/{DATASET}_ground_truth_sites.json'
GT_KLIFS = f'data/{DATASET}_klifs_pocket_sites.json'
print('runner ready')
```

## 8. Score every readout

The published readouts are re-scored in the same run, so the comparison is never across
runs or devices. Outputs are named by model, so nothing overwrites anything.

```python
OUT_R = f'{WORK}/readouts_{DATASET}'
OUT_R_KLIFS = f'{OUT_R}_klifs'
for d in (OUT_R, OUT_R_KLIFS):
    os.makedirs(d, exist_ok=True)

GROUND_TRUTHS = []
if RUN_UNIPROT:
    GROUND_TRUTHS.append(('uniprot', GT, OUT_R))
if RUN_KLIFS:
    GROUND_TRUTHS.append(('klifs', GT_KLIFS, OUT_R_KLIFS))

ALL_READOUTS = list(PUBLISHED) + list(VARIANTS)
jobs = []
for label, gt, folder in GROUND_TRUTHS:
    for m in ALL_READOUTS:
        tag = f'{m}_{DATASET}' if m != 'coldsite_dti' else DATASET
        for s in SEEDS:
            jobs.append((f'{m} seed {s}: {label} ladder',
                         py('src.evaluation.run_ladder', '--model', m, '--task', 'binary',
                            '--dataset', DATASET, '--seed', s, *COMMON,
                            '--ground-truth', gt, '--out-dir', folder, '--device', 'cuda'),
                         os.path.join(folder, f'ladder_{tag}_seed{s}.json')))

# Faithfulness does not depend on the ground truth, so it is built once -- outside the
# loop above. Inside it (with a break) it silently cancelled the KLIFS ladders.
if RUN_FAITHFULNESS:
    for m in ALL_READOUTS:
        if base_model_name(m) == 'coldsite_dti':
            continue              # its masking does not go through residue_space
        tag = f'{m}_{DATASET}'
        for s in SEEDS:
            jobs.append((f'{m} seed {s}: faithfulness',
                         py('src.evaluation.run_faithfulness', '--model', m,
                            '--dataset', DATASET, '--seed', s, '--task', 'binary',
                            *COMMON, '--out-dir', OUT_R, '--max-pairs', 200,
                            '--device', 'cuda'),
                         os.path.join(OUT_R, f'faithfulness_{tag}_seed{s}.json')))

print(f'{len(jobs)} jobs')
if jobs:
    cut = run_parallel(spread(jobs), 'readouts')
    print()
    print('CUT SHORT by the deadline' if cut else 'every readout scored')
```

## 9. Do the readouts agree?

Two things to read. **Verdict** is what the paper claims: at chance against UniProt's
residues, above chance against the pocket. **Overlap** is how many of the top ten residues
a variant shares with the published readout — if that is near zero while both score the
same, the residues shown in a published attention figure are not the model's answer but
the readout's.

```python
import pandas as pd

K = '10'
LEVELS = ('random', 'cold_drug', 'cold_target', 'cold_pair')


def at_k(folder, model, seed, level, field='precision_at_k'):
    tag = f'{model}_{DATASET}' if model != 'coldsite_dti' else DATASET
    path = os.path.join(folder, f'ladder_{tag}_seed{seed}.json')
    if not os.path.exists(path):
        return None
    return json.load(open(path)).get(level, {}).get('by_k', {}).get(K, {}).get(field)


def mean_sd(values):
    # plain arithmetic: statistics.* raises on some value types under Kaggle's Python 3.12
    xs = [float(v) for v in values if v is not None]
    if not xs:
        return '—'
    mean = sum(xs) / len(xs)
    if len(xs) < 2:
        return f'{mean:.3f}'
    sd = (sum((x - mean) ** 2 for x in xs) / (len(xs) - 1)) ** 0.5
    return f'{mean:.3f} ± {sd:.3f}'


rows = []
for label, _gt, folder in GROUND_TRUTHS:
    for m in ALL_READOUTS:
        base = base_model_name(m)
        for level in LEVELS:
            chance = [at_k(folder, m, s, level, 'chance') for s in SEEDS]
            chance = [c for c in chance if c is not None]
            rows.append({
                'ground truth': label, 'model': base,
                'readout': 'published' if m in PUBLISHED else m[len(base) + 1:],
                'level': level,
                'chance': f'{sum(chance) / len(chance):.3f}' if chance else '—',
                f'precision@{K}': mean_sd([at_k(folder, m, s, level) for s in SEEDS]),
            })

frame = pd.DataFrame(rows)
print(frame.to_string(index=False))
frame.to_csv(f'{WORK}/readout_comparison.csv', index=False)
print()
print(f'Saved -> {WORK}/readout_comparison.csv')
print('A readout that moves a number is the sensitivity Limitations promises to report; '
      'one that flips a verdict -- chance to above-chance, or the reverse -- is a finding '
      'about how much an attention figure depends on how it was drawn.')
```
